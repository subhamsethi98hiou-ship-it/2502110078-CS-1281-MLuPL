```python
def inspect(dia):
    """Return bin label for a measured diameter."""
    if dia < 9.90 or dia > 10.10:
        return "scrap"
    elif dia < 9.95 or dia > 10.05:
        return "rework"
    else:
        return "accept"


def tally(labels):
    """Return counts of accept, rework and scrap labels."""
    counts = {"accept": 0, "rework": 0, "scrap": 0}

    for label in labels:
        if label in counts:
            counts[label] += 1

    return counts


def report(shift, counts):
    """Return a formatted inspection report line."""
    return f"{shift}: accept={counts['accept']}, rework={counts['rework']}, scrap={counts['scrap']}"


# Test
labels = []

for dia in [10.00, 9.93, 10.07, 9.85, 10.03]:
    labels.append(inspect(dia))

counts = tally(labels)
print(report("DAY", counts))
```

```python
def build_catalogue(lines):
    """Build a part-code to diameter dictionary from log lines."""
    catalogue = {}

    for line in lines:
        parts = line.split()
        if len(parts) >= 2:
            code = parts[0]
            diameter = float(parts[1])
            catalogue[code] = diameter

    return catalogue


def lookup(catalogue, code):
    """Return the diameter for a code or a sensible default."""
    return catalogue.get(code, 0.0)


def scrap_codes(catalogue):
    """Return part codes whose diameters are outside specification."""
    result = set()

    for code, diameter in catalogue.items():
        if diameter < 9.90 or diameter > 10.10:
            result.add(code)

    return result


# Test
lines = [
    "P01 10.00",
    "P02 9.93",
    "P03 10.08",
    "P04 9.85",
    "P05 10.12"
]

catalogue = build_catalogue(lines)

print(catalogue)
print(lookup(catalogue, "P01"))
print(lookup(catalogue, "P99"))
print(scrap_codes(catalogue))
```

```python
from shoptools import inspect, tally, report

def run_inspection(data, shift):
    """Inspect all diameters and return the shift report."""
    labels = [inspect(dia) for dia in data]
    counts = tally(labels)
    return report(shift, counts)


data = [10.00, 9.93, 10.07, 9.85, 10.03]

print(run_inspection(data, "DAY"))
```
