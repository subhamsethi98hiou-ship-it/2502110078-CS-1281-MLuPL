```python
%%writefile shoptools.py

class PressMachine:

    def __init__(self, name, rating, limit):
        self.name = name
        self.rating = rating
        self.limit = limit
        self.__state = "stopped"
        self.parts = 0

    def __str__(self):
        return f"{self.name} | Rating: {self.rating} | Limit: {self.limit}"

    def start(self):
        if self.__state == "running":
            return "Machine is already running."
        self.__state = "running"
        return "Machine started."

    def stop(self):
        if self.__state == "stopped":
            return "Machine is already stopped."
        self.__state = "stopped"
        return "Machine stopped."

    def operate(self, quantity):
        if quantity < 0 or quantity > self.limit:
            return f"Rejected: quantity must be between 0 and {self.limit}."

        if self.__state != "running":
            return "Rejected: machine is not running."

        self.parts += quantity
        return f"{quantity} parts processed."

    def status(self):
        return {
            "name": self.name,
            "state": self.__state,
            "parts": self.parts,
            "rating": self.rating,
            "limit": self.limit
        }
```

```python
import random


class Machine:

    def __init__(self, name, rating, machine_id):
        self.name = name
        self.rating = rating
        self.machine_id = machine_id
        self.__state = "STOPPED"
        self.completed = 0
        self.rejected = 0

    # IT12: String representation
    def __str__(self):
        return f"{self.name} (ID: {self.machine_id}, Rating: {self.rating})"

    # IT12: Nameplate
    def nameplate(self):
        return {
            "name": self.name,
            "machine_id": self.machine_id,
            "rating": self.rating
        }

    # IT12: Working method
    def working_method(self):
        return "Operates within rated capacity."

    # IT10: Start machine
    def start(self):
        if self.__state == "RUNNING":
            print("Error: Machine is already running.")
            return

        self.__state = "RUNNING"
        print(self.name, "started.")

    # IT10: Stop machine
    def stop(self):
        if self.__state == "STOPPED":
            print("Error: Machine is already stopped.")
            return

        self.__state = "STOPPED"
        print(self.name, "stopped.")

    # IT10: Status getter
    def status(self):
        return self.__state

    # IT8: Process jobs
    def operate(self, value):

        if self.__state != "RUNNING":
            print("Error: Start machine first.")
            return False

        if value < 0 or value > self.rating:
            self.rejected += 1
            print("Out of range:", value)
            return False

        self.completed += 1
        print("Job completed:", value)
        return True


# IT12: Create independent machines
machine1 = Machine("Lathe", 100, "M101")
machine2 = Machine("Drill", 80, "M102")

print("\n--- IT12: MACHINE DETAILS ---")
print(machine1)
print(machine2)

print(machine1.nameplate())
print(machine2.nameplate())

print(machine1.working_method())
print(machine2.working_method())

print("Independent machines:", machine1 is not machine2)


# IT10: Guard testing
print("\n--- IT10: GUARDED MACHINE ---")

machine1.start()
machine1.operate(50)
machine1.operate(150)
machine1.stop()

print("Machine status:", machine1.status())

# Private state bypass test
machine1.__state = "RUNNING"
print("External state:", machine1.__state)
print("Actual state:", machine1.status())


# IT8: Batch processing
print("\n--- IT8: BATCH PROCESSING ---")

machine1.start()

jobs = [20, 30, 110, 40, 50, 120, 60, 70, 80, 150]

for job in jobs:
    machine1.operate(job)

machine1.stop()

print("\nSHIFT REPORT")
print("Completed:", machine1.completed)
print("Rejected:", machine1.rejected)


# IT6: Fleet management
print("\n--- IT6: FLEET MANAGEMENT ---")

fleet = [
    Machine("Lathe", 100, "M101"),
    Machine("Drill", 80, "M102"),
    Machine("Milling", 120, "M103")
]

jobs = [50, 90, 70, 110, 60, 130, 40, 80, 100, 30]

for job in jobs:

    allocated = False

    for machine in fleet:

        if machine.status() == "STOPPED" and job <= machine.rating:

            machine.start()
            machine.operate(job)
            machine.stop()

            allocated = True
            break

    if not allocated:
        print("No suitable machine for job:", job)


# Final plant report
print("\n--- FINAL PLANT REPORT ---")

total = 0

for machine in fleet:

    print("\nMachine:", machine.name)
    print("Completed:", machine.completed)
    print("Rejected:", machine.rejected)

    total += machine.completed

print("\nTotal completed jobs:", total)
print("Total machines:", len(fleet))
```
