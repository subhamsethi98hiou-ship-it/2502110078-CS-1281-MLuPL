```python
from pathlib import Path

def load_jobs(file):
    try:
        with open(file) as f:
            return [line.strip() for line in f if line.strip()]
    except FileNotFoundError:
        print("Error: Jobs file not found!")
        return []
    except PermissionError:
        print("Error: Permission denied!")
        return []
```

```python
class Machine:
    def __init__(self, name, capacity, status):
        self.name = name
        self.capacity = capacity
        self.status = status

    def start(self):
        self.status = 1

    def stop(self):
        self.status = 0


def run_tests():
    m = Machine("Test", 100, 0)

    assert m.status == 0
    m.start()
    assert m.status == 1
    m.stop()
    assert m.status == 0

    print("All tests passed!")


if __name__ == "__main__":
    run_tests()
```
