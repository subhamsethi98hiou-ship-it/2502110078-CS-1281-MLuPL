**IT8**

```python
log_lines = [
    " PE-26-0137 ; 25.02 ; DAY ",
    "PE-26-0138; 24.97; DAY",
    " PE-26-0139 ; 25.05 ; DAY ",
    "PE-26-0140; 25.01; NIGHT",
    " PE-26-0141 ; 24.95 ; NIGHT ",
    "PE-26-0142 ; 25.03 ; DAY",
    "PE-26-0143; 25.08; NIGHT",
    " PE-26-0144 ; 24.99 ; DAY ",
    "PE-26-0145 ; 25.06 ; NIGHT",
    " PE-26-0146 ; 25.00 ; NIGHT "
]

accepted = []
rework = []
scrap = []

for line in log_lines:
    fields = line.strip().split(";")

    code = fields[0].strip()
    dia = float(fields[1].strip())
    shift = fields[2].strip().upper()

    if 24.98 <= dia <= 25.05:
        accepted.append((code, dia, shift))

    elif 24.95 <= dia <= 25.08:
        rework.append((code, dia, shift))

    else:
        scrap.append((code, dia, shift))


print("Accepted:", len(accepted))
print("Rework:", len(rework))
print("Scrap:", len(scrap))


accepted_diameters = [item[1] for item in accepted]

if accepted_diameters:
    accepted_average = sum(accepted_diameters) / len(accepted_diameters)
    print("Accepted Average:", accepted_average)


print("Accepted Parts:")
for item in accepted:
    print(item)


print("Rework Parts:")
for item in rework:
    print(item)


print("Scrap Parts:")
for item in scrap:
    print(item)
```

```python
log_lines = [
    " PE-26-0137 ; 25.02 ; DAY ",
    "PE-26-0138; 24.97; DAY",
    " PE-26-0139 ; 25.05 ; DAY ",
    "PE-26-0140; 25.01; NIGHT",
    " PE-26-0141 ; 24.95 ; NIGHT ",
    "PE-26-0142 ; 25.03 ; DAY",
    "PE-26-0143; 25.08; NIGHT",
    " PE-26-0144 ; 24.99 ; DAY ",
    "PE-26-0145 ; 25.06 ; NIGHT",
    " PE-26-0146 ; 25.00 ; NIGHT "
]

day_accepted = []
day_scrap = []

night_accepted = []
night_scrap = []

for line in log_lines:

    fields = line.strip().split(";")

    code = fields[0].strip()
    dia = float(fields[1].strip())
    shift = fields[2].strip().upper()


    if 24.98 <= dia <= 25.05:
        status = "ACCEPTED"

    elif 24.95 <= dia <= 25.08:
        status = "REWORK"

    else:
        status = "SCRAP"



    if shift == "DAY":

        if status == "ACCEPTED":
            day_accepted.append(code)

        elif status == "SCRAP":
            day_scrap.append(code)

    elif shift == "NIGHT":

        if status == "ACCEPTED":
            night_accepted.append(code)

        elif status == "SCRAP":
            night_scrap.append(code)


day_total = 0
night_total = 0

for line in log_lines:

    fields = line.strip().split(";")
    shift = fields[2].strip().upper()

    if shift == "DAY":
        day_total += 1

    elif shift == "NIGHT":
        night_total += 1


day_accept_rate = (len(day_accepted) / day_total) * 100
night_accept_rate = (len(night_accepted) / night_total) * 100


print("DAY ACCEPT RATE:", day_accept_rate, "%")
print("NIGHT ACCEPT RATE:", night_accept_rate, "%")


day_scrap_set = set(day_scrap)
night_scrap_set = set(night_scrap)

both_scrap = day_scrap_set & night_scrap_set
day_only_scrap = day_scrap_set - night_scrap_set
night_only_scrap = night_scrap_set - day_scrap_set


print("Scrapped in BOTH shifts:", both_scrap)
print("Scrapped only in DAY:", day_only_scrap)
print("Scrapped only in NIGHT:", night_only_scrap)
```
