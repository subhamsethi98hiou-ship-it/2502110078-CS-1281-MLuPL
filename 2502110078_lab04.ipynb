**IT10**(loop 10)

```python
import random

accept = 0
rework = 0
scrap = 0

for i in range(10):
    d = random.uniform(24.90, 25.10)
    d = round(d, 2)

    if 24.95 <= d <= 25.05:
        accept += 1
        print(d, "ACCEPT")

    elif 24.90 <= d < 24.95 or 25.05 < d <= 25.10:
        rework += 1
        print(d, "REWORK")

    else:
        scrap += 1
        print(d, "SCRAP")

print("ACCEPT:", accept)
print("REWORK:", rework)
print("SCRAP:", scrap)
```

**IT8**

```python
import random

mode = input("Enter mode (halt/skip): ").lower()

accept = 0
rework = 0
scrap = 0

for i in range(100):
    d = round(random.uniform(24.90, 25.10), 2)

    if 24.95 <= d <= 25.05:
        accept += 1
        print(d, "ACCEPT")

    elif 24.90 <= d < 24.95 or 25.05 < d <= 25.10:
        rework += 1
        print(d, "REWORK")

    else:
        scrap += 1
        print(d, "SCRAP")

        if mode == "halt":
            break

print("ACCEPT:", accept)
print("REWORK:", rework)
print("SCRAP:", scrap)
print("Parts inspected:", accept + rework + scrap)
```

**IT6**

```python
import random

consecutive_accepts = 0
total_parts = 0

while consecutive_accepts < 5 and total_parts < 500:

    d = round(random.uniform(24.90, 25.10), 2)
    total_parts += 1

    if 24.95 <= d <= 25.05:
        consecutive_accepts += 1
        print(d, "ACCEPT",
              "Consecutive:", consecutive_accepts)

    else:
        consecutive_accepts = 0
        print(d, "NOT ACCEPT")

print("Total parts inspected:", total_parts)
print("Consecutive ACCEPTS:", consecutive_accepts)
```
