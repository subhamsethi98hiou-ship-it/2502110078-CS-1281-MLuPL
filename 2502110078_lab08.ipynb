```python
def parse_report(data):
    records = []
    rejects = []

    for line_no, line in enumerate(data, 1):
        try:
            parts = line.split(",")

            if len(parts) != 3:
                raise ValueError("wrong number of fields")

            name = parts[0].strip()

            if not name:
                raise ValueError("missing name")

            hours = float(parts[1].strip())

            if hours < 0:
                raise ValueError("negative hours")

            job = parts[2].strip()

            if not job:
                raise ValueError("missing job")

            records.append({
                "name": name,
                "hours": hours,
                "job": job
            })

        except ValueError as e:
            rejects.append({
                "line": line_no,
                "reason": str(e)
            })

    return records, rejects


data = [
    "Amit,8,Job1",
    "Riya,7,Job2",
    "Wrong,abc,Job3",
    "Rahul,9,Job3"
]

records, rejects = parse_report(data)

print("Records:")
print(records)

print("Rejects:")
print(rejects)
```

**IT8**

```python
class SpecViolation(Exception):
    pass


def operate(line):
    parts = line.split(",")

    # Check number of fields
    if len(parts) != 3:
        raise SpecViolation("wrong number of fields")

    # Get name
    name = parts[0].strip()

    if not name:
        raise SpecViolation("missing name")

    # Convert hours safely
    try:
        hours = float(parts[1].strip())
    except ValueError:
        raise SpecViolation("hours must be a number")

    # Check hours
    if hours < 0 or hours > 24:
        raise SpecViolation("invalid working hours")

    # Get job
    job = parts[2].strip()

    if not job:
        raise SpecViolation("missing job")

    return {
        "name": name,
        "hours": hours,
        "job": job
    }


def run_program(data):
    records = []
    rejects = []

    for line_no, line in enumerate(data, 1):
        try:
            record = operate(line)
            records.append(record)

        except SpecViolation as e:
            rejects.append({
                "line": line_no,
                "reason": str(e)
            })

    return records, rejects


# Test data
data = [
    "Amit,8,Job1",
    "Riya,7,Job2",
    "Bad,abc,Job3",
    "Rahul,30,Job4",
    "Sita,-2,Job5",
    "Raj,10,Job6"
]


# Run program
records, rejects = run_program(data)

print("Records:", records)
print("Rejects:", rejects)
```

```python
def run_tests():
    tests = [
        "Amit,8,Job1",
        "Riya,7,Job2",
        "Bad,abc,Job3",
        "Rahul,-2,Job4",
        ",8,Job5",
        "Sita,10,",
        "Wrong,8",
        "Raj,30,Job8"
    ]

    passed = 0

    for i, test in enumerate(tests, 1):
        try:
            operate(test)
            result = "PASS"
            passed += 1
        except (SpecViolation, ValueError):
            result = "PASS"

        print("Test", i, ":", result)

    print("\nPassed:", passed, "/", len(tests))


run_tests()
```
