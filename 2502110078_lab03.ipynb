**IT10**

```python
accept = 0
rework = 0
scrap = 0

for i in range(10):
    d = float(input("Enter diameter:"))

    if 9.95 <= d <= 10.05:
        print("ACCEPT")
        accept += 1

    elif 9.90 <= d < 9.95 or 10.05 < d <= 10.10:
        print("REWORK")
        rework += 1

    else:
        print("SCRAP")
        scrap += 1

print("Accept =", accept)
print("Rework =", rework)
print("Scrap =", scrap)
```

**IT8**

```python
total = int(input("Enter total parts: "))
size = int(input("Enter box size: "))

if size == 0:
    print("Box size cannot be 0")
else:
    boxes = total // size
    loose = total % size

    print("Full boxes:", boxes)
    print("Loose parts:", loose)
    print(boxes * size + loose == total)
```

```python
accept = 0
rework = 0
scrap = 0

for i in range(10):
    d = float(input("Enter diameter: "))

    if 9.95 <= d <= 10.05:
        accept += 1
    elif 9.90 <= d < 9.95 or 10.05 < d <= 10.10:
        rework += 1
    else:
        scrap += 1

boxes = accept // 4
loose = accept % 4

print("Accepted:", accept)
print("Rework:", rework)
print("Scrap:", scrap)
print("Boxes:", boxes)
print("Loose accepted parts:", loose)
```
